```python

name=['Bill','Wes','John','Jake']    # Initial list of students along with their marks
Science=[43,45,37,48]
Math=[47,48,49,45]
Total= [90,93,86,93]
students={'name': name, 'Science': Science, 'Math': Math,'Total':Total}     # Dictionary - Key:Value pair

with open("students_record.txt", "w") as file:   # Writing the data into students_record file, using 'w' command
    file.write("========== STUDENT RECORDS ==========\n")
    file.write("-" * 38 + "\n")

    for i in range(len(students['name'])):
        file.write(f"Name:    {students['name'][i]}\n")
        file.write(f"Science: {students['Science'][i]}\n")
        file.write(f"Math:    {students['Math'][i]}\n")
        file.write(" " * 38 + "\n")
        file.write(f"Total:   {students['Total'][i]}\n")
        file.write("-" * 38 + "\n")     # Lines to divide one student from another, making it easily readable. 
        
with open("students_record.txt", "r") as file:  # Reading the data from students_record file, using 'r' command
    content = file.read()
    print(content)


while True:    # Keeps running, so the user can try out all the options unless the user want to exit the option
    option = int(input("""\n -------------------------
                       \n Enter the option: 
                       \n -------------------------
                       \n1.Number of students 
                       \n2.Best performer in Science 
                       \n3.Average marks in Science 
                       \n4.Best performer in Math
                       \n5.Average marks in Math 
                       \n6. Add a student
                       \n7. Best overall performer in both subjects
                       \n8. Specific student marks
                       \n9. Exit options
                       \n -------------------------"""))

    if option==1: # Count of students
        print(f"There are {len(students['name'])} students in this record.")    # f-string, to make the message dynamic. 
        
        with open("students_record.txt", "r") as file:
            content = file.read()
            print(content)
        
    elif option==2: 
        for i in range(len(students['Science'])):
            if students['Science'][i] == max(students['Science']):
                print(f"Best Science performer is {students['name'][i]} with {max(students['Science'])} marks")
    elif option==3:
        print(f"Average marks in Science is {sum(students['Science'])/len(students['Science'])}")   
    elif option==4:
        for i in range(len(students['Science'])):
            if students['Math'][i] == max(students['Math']):
                print(f"Best Math performer is {students['name'][i]} with {max(students['Math'])} marks")
    elif option==5:
        print(f"Average marks in Math is {sum(students['Math'])/len(students['Math'])}")   
    elif option==6:
        try:   # try and except, for exception handling, which prevents the program from crashing when an unexpected error occurs at runtime.
            t=int(input('how many students you want to add: '))
            start = len(students['name'])   # So the range can start from the last enter value instead the first value. 

            if t==0:
                pass   # if 0 is selected, it skips to the next option without throwing an error
            else:
                for i in range(t):     # Keeps running for number of students
                    
                    name = input("Enter student name: ")
                    if name.isdigit():
                        print("Please enter a valid name.")
                    else:
                        science = int(input("Enter Science marks for 50: "))
                        math = int(input("Enter Math marks for 50: "))

                        students['name'].append(name)
                        students['Science'].append(science)
                        students['Math'].append(math)
                        total=science+math
                        students['Total'].append(total)
        except ValueError:
            print("Please enter a number for marks.")
  
        with open("students_record.txt", "a") as file:     # Joins/ Appends the new student's name/marks to the record
            for i in range(start, len(students['name'])):    # Implementation of 'start'
                file.write(f"Name:    {students['name'][i]}\n")
                file.write(f"Science: {students['Science'][i]}\n")
                file.write(f"Math:    {students['Math'][i]}\n")
                file.write(" " * 38 + "\n")
                file.write(f"Total:   {students['Total'][i]}\n")
                file.write("-" * 38 + "\n")
        
        
    
    elif option==7:    # Filters the marks and then maps the name to the marks and gives us a list of students
        
        print(f"Best overall performers list with atleast 90 marks is {(list(map(lambda i: students['name'][i],filter(lambda i: students['Total'][i] >= 90, range(len(students['Total']))))))}")
    
    elif option == 8:
        s = input("Enter the student name: ").title()   # capitalizes the first character of every word and changes all remaining letters to lowercase

        if s in students['name']:
            i = students['name'].index(s)     # Position of the student to find all the marks and the total

            print(f"{s}'s marks are:")
            print(f"Science: {students['Science'][i]}")
            print(f"Math: {students['Math'][i]}")
            print(f"Total: {students['Total'][i]}")
          
        else:
            print("Student not found")   # Message instead of throwing an error

        
    elif option==9:   # Breaks the loops, by exiting the selection
        print('Exiting the option section')
        break
    
    else:    # Gives a message if a number out the options is selected without throwing an error.
        print('Enter a valid option')

   
```
